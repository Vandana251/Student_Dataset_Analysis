MATPLOTLIB


```python
plt.plot(df["StudyHours"], df["Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
plt.scatter(df["StudyHours"], df["Marks"])
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
plt.hist(df["Marks"].dropna(), bins=10)
plt.xlabel("Marks")
plt.ylabel("Number of Students")
plt.title("Marks Distribution")
plt.show()
```

```python
branch_count = df["Branch"].value_counts()

plt.bar(branch_count.index, branch_count.values)
plt.xlabel("Branch")
plt.ylabel("Number of Students")
plt.title("Students by Branch")
plt.show()
```

```python
plt.scatter(df["AttendancePercent"], df["Marks"])
plt.xlabel("Attendance Percentage")
plt.ylabel("Marks")
plt.title("Attendance vs Marks")
plt.show()
```

SEABORN

```python
sns.scatterplot(data=df, x="StudyHours", y="Marks")
plt.title("Study Hours vs Marks")
plt.show()
```

```python
sns.histplot(data=df, x="Marks")
plt.title("Marks Distribution")
plt.show()
```

```python
sns.boxplot(data=df, y="Marks")
plt.title("Marks Distribution")
plt.show()
```

```python
sns.boxplot(data=df, x="Branch", y="Marks")
plt.title("Marks by Branch")
plt.show()
```

```python
sns.boxplot(data=df, x="Gender", y="Marks")
plt.title("Marks by Gender")
plt.show()
```

```python
sns.regplot(data=df, x="StudyHours", y="Marks")
plt.title("Study Hours vs Marks")
plt.show()
```
